## The Data

This classic dataset contains the prices and other attributes of almost 54,000 diamonds. It's a great dataset for beginners learning to work with data analysis and visualization.

#### Content

* price price in US dollars (\$326--\$18,823)

* carat weight of the diamond (0.2--5.01)

* cut quality of the cut (Fair, Good, Very Good, Premium, Ideal)

* color diamond colour, from J (worst) to D (best)

* clarity a measurement of how clear the diamond is (I1 (worst), SI2, SI1, VS2, VS1, VVS2, VVS1, IF (best))

* x length in mm (0--10.74)

* y width in mm (0--58.9)

* z depth in mm (0--31.8)
 
* depth total depth percentage = z / mean(x, y) = 2 * z / (x + y) (43--79)

* table width of top of diamond relative to widest point (43--95)

## Imports and Data

#### TASK: Create a scatterplot of price versus carat as shown below

```python
plt.figure(figsize=(4, 3))
sns.scatterplot(x='carat', y='price', data=diamonds)
plt.show()
```

#### TASK: Create a histogram of the price column as shown below

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=diamonds, x='price', bins=40, edgecolor='lightblue', color='lightblue')
plt.show()
```

#### TASK: Create a count plot of the instances per cut type as shown below

```python
plt.figure(figsize=(8, 6))

cut_order = ['Ideal', 'Premium', 'Very Good', 'Good', 'Fair']
sns.countplot(data=diamonds, x='cut', order=cut_order, palette='Set2')

plt.title('Count Plot of Diamond Cuts')
plt.xlabel('Cut Quality')
plt.ylabel('Count')
plt.show()
# warning biarin aja
```

#### TASK: Create a large box plot figure showing the price distribution per cut type as shown below

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=diamonds, x='cut', y='price', palette='Set2')

plt.title('Harga distribusi per potongan')
plt.xlabel('Potongan')
plt.ylabel('Harga')
plt.show()
```
